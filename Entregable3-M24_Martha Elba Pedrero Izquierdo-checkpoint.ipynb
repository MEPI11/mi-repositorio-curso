### Cargar las tablas de datos en Pandas:

```python
# se llaman los directorios
import pandas as pd
import numpy as np
```

```python
# Se llaman los archivos, 
dim_category = pd.read_csv("DIM_CATEGORY.csv", encoding = "latin-1")
dim_product  = pd.read_csv("DIM_PRODUCT.csv", encoding = "latin-1")
dim_segment  = pd.read_csv("DIM_SEGMENT.csv", encoding = "latin-1")
dim_calendar = pd.read_csv("DIM_CALENDAR.csv", encoding = "latin-1")
fact_sales   = pd.read_csv("FACT_SALES.csv", encoding = "latin-1")
```

###  Revisar y entender los datos cargados:

```python
#Revisión rápida de las primeras filas de cada DataFrame se usa head(), info() y describe() 

def explorar_df(nombre, df):
    print("1a fila (head):")
    display(df.head())
    print("Estructira(info):")
    df.info()
    print("Descripcion(describe):")
    display(df.describe(include='all'))

explorar_df("DIM_CATEGORY", dim_category)
explorar_df("DIM_PRODUCT", dim_product)
explorar_df("DIM_SEGMENT", dim_segment)
explorar_df("DIM_CALENDAR", dim_calendar)
explorar_df("FACT_SALES", fact_sales)
```

###  Realizar limpieza de datos:

```python
# se eliminan los duplicados
dim_category = dim_category.drop_duplicates()
dim_product  = dim_product.drop_duplicates()
dim_segment  = dim_segment.drop_duplicates()
dim_calendar = dim_calendar.drop_duplicates()
fact_sales   = fact_sales.drop_duplicates()
```

```python
#Se quitan de nulos o  Valores faltantes: Si Non-Null Count es menor al total de filas, hay NaN, al revisar la estructura de los datos se encontro:
# de los 5 archivos solo dim_product presenta nulos y esto se sabe por la estructura, pues dice que deben ser 505 valores
# pero no los tienen los siguientes elementos ITEM (503), ATTR1 (499) y ATTR3 (499)
dim_product['ITEM'] = dim_product['ITEM'].fillna('SIN DATOS')
dim_product['ATTR1'] = dim_product['ATTR1'].fillna('SIN DATOS')
dim_product['ATTR3'] = dim_product['ATTR3'].fillna('SIN DATOS')

```

```python
# limpiar las celdas
# se quitan los espacios antes y despúes del texto, se cambia para que todas las letras sean minusculas
def limpieza(df):
    columnas_limpias = df.select_dtypes(include='object').columns
    for col in columnas_limpias:
        df[col] = df[col].astype(str).str.strip().str.title()
    return df
dim_category = limpieza(dim_category)
dim_product  = limpieza(dim_product)
dim_segment  = limpieza(dim_segment)
dim_calendar = limpieza(dim_calendar)
```

### Unir los DataFrames relevantes:

```python
df_consolidado = fact_sales.copy()

df_consolidado = df_consolidado.merge(
    dim_product, 
    left_on='ITEM_CODE',    # en fact_sales
    right_on='ITEM',        # en dim_product
    how='left')

df_consolidado = df_consolidado.merge(
    dim_calendar, on='WEEK', how='left')

print(f"Final-1: {df_consolidado.shape}")
```

```python
df_consolidado2 = dim_segment.copy()

df_consolidado2['CATEGORY'] = df_consolidado2['CATEGORY'].astype(str)
dim_category['CATEGORY'] = dim_category['CATEGORY'].astype(str)

df_consolidado2 = df_consolidado2.merge(
    dim_category, on='CATEGORY', how='left')
print(f"Final-2: {df_consolidado2.shape}")
```

### Aplicar transformaciones necesarias:

```python
#Estandarizar fechas 
for df in [dim_calendar, fact_sales]:
    if 'date' in df.columns:
        df['date'] = pd.to_datetime(df['date'], format='mixed', errors='coerce')

# Corregir formatos numéricos
    fact_sales['TOTAL_VALUE_SALES'] = (fact_sales['TOTAL_VALUE_SALES']
        .astype(str).str.replace(r'[\$,€]', '', regex=True).str.replace(',', '.').astype(float))
```

### Guardar el conjunto de datos consolidado:

```python
RUTA_FINAL = "dataset_consolidado_final.csv"
df_consolidado.to_csv(RUTA_FINAL, index=False, encoding='utf-8')

RUTA_FINAL2 = "dataset_consolidado_final-2.csv"
df_consolidado2.to_csv(RUTA_FINAL2, index=False, encoding='utf-8')
```
